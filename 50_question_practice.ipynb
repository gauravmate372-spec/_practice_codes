```python
#Q5
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
```

```python
#Q6
def fibonacci(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(6))
```

```python
#Q9
x = 10

def outer():
    x = 20

    def inner():
        nonlocal x
        x += 5

    inner()
    return x

print(outer())
```

```python
#Q11
def decorator(func):
    def wrapper():
        print("Starting")
        func()
        print("Finished")
    return wrapper

@decorator
def hello():
    print("Hello")

hello()
```

```python
#Q12
def decorator(func):
    def wrapper():
        print("Before")
        func()
        print("After")
    return wrapper

@decorator
def greet():
    print("Hello")

greet()
```

```python
#Q13
def deco(fn):
    def wrap():
        print("A")
        fn()
        print("B")
    return wrap

@deco
def hello():
    print("Hi")

hello()
```

```python
#Q14
import time
def timer(func):
    def wrapper():
        start = time.perf_counter()
        func()
        end = time.perf_counter()
        print("Time taken:", end - start)

    return wrapper

@timer
def work():
    print("Working")

work()
```

```python
#Q15
def decorator(func):
    def wrapper(*args, **kwargs):
        print("Function called")
        return func(*args, **kwargs)
    return wrapper

@decorator
def add(a, b):
    return a + b

print(add(10, 20))
```

```python
#Q16
from functools import wraps

def decorator(func):
    @wraps(func)
    def wrapper():
        return func()
    return wrapper

@decorator
def greet():
    print("Hello")

print(greet.__name__)
```

```python
#Q17
def admin_only(func):
    def wrapper(user_is_admin):
        if user_is_admin:
            return func(user_is_admin)
        print("Access denied")
    return wrapper

@admin_only
def open_panel(user_is_admin):
    print("Admin panel opened")

open_panel(True)
open_panel(False)
```

```python
#Q18
def retry(func):
    def wrapper():
        for attempt in range(3):
            try:
                return func()
            except Exception:
                print("Retrying...")
        print("Failed after 3 attempts")
    return wrapper

@retry
def test():
    raise ValueError("Error")

test()
```

```python
#Q20
def gen():
    yield 1
    yield 2
    return 3

g = gen()

print(next(g))
print(next(g))

try:
    print(next(g))
except StopIteration as e:
    print("Done", e.value)
```

```python
#Q21
def first():
    return 10

def second():
    yield 10

print(first())
print(next(second()))
```

```python
#Q22
def even_numbers(limit):
    for number in range(0, limit + 1):
        if number % 2 == 0:
            yield number

print(list(even_numbers(10)))
```

```python
#Q24
def squares(n):
    for number in range(n):
        yield number * number

for value in squares(4):
    print(value)
```
