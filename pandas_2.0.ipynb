```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\Admin\Documents\industry_retail_orders_dirty_1000.csv")
```

```python
df.info()
```

```python
# AS In this dataset you can see there are 12 column that are having null values we will fill this values 
```

1. customer age

```python
df["customer_age"] = df["customer_age"].fillna(df["customer_age"].mean())
df
```

2. Unit_price

```python
df["unit_price"] = df["unit_price"].fillna(df["unit_price"].median())
df
```

3. discount_pct

```python
df["discount_pct"] = df["discount_pct"].fillna(df["discount_pct"].mean())
df
```

4. shipping_fee

```python
df["shipping_fee"] = df["shipping_fee"].fillna(df["shipping_fee"].median())
df
```

5. delivery_days

```python
df["delivery_days"] = df["delivery_days"].fillna(df["delivery_days"].median())
df
```

6.customer_rating

```python
df["customer_rating"] = df["customer_rating"].fillna(df["customer_rating"].mean())
df
```

7. customer_segment

```python
df["customer_segment"] = df["customer_segment"].fillna(df["customer_segment"].mode()[0])
df
```

8. city

```python
df["city"] = df["city"].fillna(df["city"].mode()[0])
df
```

9. product_category

```python
df["product_category"] = df["product_category"].fillna(df["product_category"].mode()[0])
df
```

10. payment_method

```python
df["payment_method"] = df["payment_method"].fillna("Unknown")
df
```

11.marketing_campaign

```python
df["marketing_campaign"] = df["marketing_campaign"].fillna("Unknown")
df
```
