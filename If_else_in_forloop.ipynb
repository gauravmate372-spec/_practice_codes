```python
for i in range(1, 11):
    if i % 2 == 0:
        print(i, 'is even')
    else:
        print(i, 'is odd')
```

```python
numbers = [10, -5, 0, 8, -2]
for num in numbers:
    if num > 0:
        print(num, 'is positive')
    elif num < 0:
        print(num, 'is negative')
    else:
        print(num, 'is zero')
```

```python
marks = [85, 42, 67, 30, 91]
for mark in marks:
    if mark >= 35:
        print(mark, '-pass')
    else:
        print(mark, '-fail')
```

```python
for i in range(1, 16):
    if i % 3 == 0:
        print(i, 'is a multiple of 3')
    else:
        print(i, 'is not a multiple of 3')
```
