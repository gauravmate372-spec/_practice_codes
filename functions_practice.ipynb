```python
#class + constructor
```

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def show(self):
        print("Name:", self.name)
        print("Age:", self.age)

s1 = Student("Gaurav", 21)

s1.show()
```

```python
#two objects
```

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def show(self):
        print("Name:", self.name)
        print("Age:", self.age)

s1 = Student("Gaurav", 21)
s2 = Student("Rahul", 22)

s1.show()
s2.show()
```
