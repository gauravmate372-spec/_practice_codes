```python
add = lambda a, b: a + b
print(add(10, 20)) #add a number using lambda function
```

```python
cube = lambda x: x**3
print(cube(4)) #cube of number
```

```python
even = lambda x: 'even' if x % 2 == 0 else'odd'
print(even(9)) #cube of number
```

```python
maximum = lambda a , b: a if a > b else b
print(maximum(15, 25)) #find the maximum number
```

```python
upper = lambda s: s.upper()
print(upper('python')) #convert the string into upper case
```

```python
length = lambda s: len(s)
print(length('python')) #find the length of string
```

```python
check = lambda x: 'positive' if x >= 0 else 'Negative'
print(check(-8)) #finding number is positive or negative
```

```python
numbers = [5, 2, 8, 1, 9]
numbers.sort(key=lambda x: x)
print(numbers) #sort the list
```

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
result = list(filter(lambda x: x % 2 == 0, numbers))
```
