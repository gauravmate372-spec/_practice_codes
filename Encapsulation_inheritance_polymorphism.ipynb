```python
class student:
    def __init__(self,name,marks):
        self.name = name
        self.__marks = marks
        def show_marks(self):
            print('marks:',self.__marks)
            def update_marks(self,marks):
                self.marks = marks
s = student('Gaurav',85)
print('name:' , s.name)
s.show_marks()

s.update_marks(95)
s.show_marks()
```

Inheritance

```python
class animal:
    def eat(self):
        print('Animal is eating')

class dog(animal):
    def bark(self):
        print('Dog is barking')
d = dog()
d.eat()
d.bark()
```

Polymorphism

```python
class animal:
    def sound(self):
        print('animal makes sound')
class Dog(animal):
    def sound(self):
        print('Dog barks')
class cat(animal):
    def sound(self):
        print('Cat meows')

d = Dog()
c = cat()

d.sound()
c.sound()
```
