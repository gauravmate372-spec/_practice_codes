```python
import pandas as pd
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/refs/heads/master/titanic.csv"
df = pd.read_csv(url)
print(df.info())
print("\nMissing values:")
print(df.isnull().sum())
```

```python
print(df.describe(include='all'))
```

```python
print('before deleting column:')
df.info(memory_usage='deep')
```

```python
df = df.drop('Cabin',axis=1)
```

```python
df
```

```python
print(df['Pclass'].unique())
```

```python
female_first_class = df[
    (df['Pclass'] == 1) &
    (df['Sex'] == 'female')
]
print(female_first_class)
```

```python
fare_limit = df['Fare'].quantile(0.95)
print('Top 5% Fare Limit:', fare_limit)
```

```python
df['Is_adult'] = df['Age'] >= 18
print(df[['Age', 'Is_adult']].head(10))
```

```python
masters = df[
    df['Name'].str.contains('master', case= False, na = False)
]
print(masters[['Name','Age','Sex']])
```

```python
first_10 = df.head(10)
print(first_10)
```

```python
first_10 = first_10.reset_index(drop=True)
print(first_10)
```

```python
median_age = df['Age'].median()
print('median Age', median_age)
```

```python
df['Age'] = df['Age'].fillna(median_age)
```

```python
print(df['Age'].isnull().sum())
```

```python
df['Sex'] = df['Sex'].map({
    'male':1,
    'female':0
})
print(df['Sex'].head())
```
