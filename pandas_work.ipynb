```python
import pandas as pd
import numpy as np

data = {
    "Employee": ["Rahul", "Aman", "Priya", "Shad",
                 "Neha", "Vikas", "Rohit", "Anjali",
                 "Karan", "Simran", "Mohit", "Riya"],

    "Department": ["IT", "HR", "IT", "Finance",
                   "HR", "IT", "Finance", "HR",
                   "IT", "Finance", "HR", "IT"],

    "City": ["Jaipur", "Delhi", "Pune", "Jaipur",
             "Mumbai", "Delhi", "Pune", "Mumbai",
             "Jaipur", "Delhi", "Pune", "Mumbai"],

    "Salary": [25000, 35000, 30000, 45000,
               40000, 55000, 50000, 32000,
               60000, 42000, 38000, 48000],

    "Experience": [1, 3, 2, 5, 4, 7, 6, 2, 8, 5, 3, 6],

    "Gender": ["Male", "Male", "Female", "Male",
               "Female", "Male", "Male", "Female",
               "Male", "Female", "Male", "Female"],

    "Performance": [75, 80, 85, 90, 70, 95,
                    88, 78, 92, 89, 76, 94]
}
df = pd.DataFrame(data)
df
```

# LEVEL-01

Calculate total salary department-wise.

Find average salary of each department.

Find maximum salary in each department.

Count employees in each department.

Find average experience city-wise.

groupby() is a Pandas method used to split data into groups based on one or more columns and 
perform aggregate operations such as sum, mean, count, minimum and maximum.

1. Calculate total salary department-wise.

```python
df.groupby("Department")["Salary"].sum()
```

2. Find average salary of each department.

```python
df.groupby("Department")["Salary"].mean()
```

3.Find maximum salary in each department.

```python
df.groupby("Department")["Salary"].max()
```

4. Count employees in each department.

```python
df.groupby("Department")["Employee"].count()
```

5. Find average experience city-wise.

```python
df.groupby("City")["Experience"].mean()
```

Aggregation is mainly used to summarize and analyze data

```python
df["Salary"].agg([
    "sum",
    "mean",
    "median",
    "min",
    "max"
])
```

```python
df.agg({
    "Salary": ["sum", "mean", "max"],
    "Employee": ["count"]
})
```

```python
df.groupby(
    ["Department", "Gender"]
)["Salary"].mean()
```

```python
df.groupby("Department").agg({
    "Salary": "sum",
    "Performance": "mean"
})
```

```python
df.groupby("Department")["Salary"].sum().sort_values(ascending=False)
```

# Level 3 – Transform, Filter & Ranking

11. Calculate department-wise average salary for every employee using transform().

12. Identify employees earning above their department's average salary.

13. Rank employees by salary within each department.

14. Display only departments having more than 3 employees using filter().

15. Find the top 2 highest-paid employees from each department.

```python
df["Dept_Avg_Salary"] = (
    df.groupby("Department")["Salary"].transform("mean")
)

print(df[["Employee", "Department", "Salary", "Dept_Avg_Salary"]])
```

```python
dept_avg = df.groupby("Department")["Salary"].transform("mean")

result = df[df["Salary"] > dept_avg]

print(result[["Employee", "Department", "Salary"]])
```

```python
df["Salary_Rank"] = (
    df.groupby("Department")["Salary"]
    .rank(ascending=False, method="dense")
)

print(df[["Employee", "Department", "Salary", "Salary_Rank"]])
```

```python
result = df.groupby("Department").filter(lambda x: len(x) > 3)
print(result)
```

```python
result = (
    df.sort_values("Salary", ascending=False)
      .groupby("Department")
      .head(2)
)

print(result)
```

```python
result = df.groupby("Department")["Salary"].sum()

print(result.sort_values(ascending=False))
```

```python
result = df.groupby("City")["Salary"].mean()

print(result.sort_values(ascending=False))
```

```python
result = df.groupby("Department")["Performance"].mean()

print(result)
```
