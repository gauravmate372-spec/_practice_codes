```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\Admin\Documents\industry_retail_orders_dirty_1000.csv")
```

```python
#1 Create a histogramm of customer age
```

```python
plt.hist(df["customer_age"].dropna(), bins=15, edgecolor="black")

plt.title("Distribution of Customer Age")
plt.xlabel("Customer Age")
plt.ylabel("Frequency")

plt.show()
```

```python
#2 create a histogram of quantity
```

```python
plt.hist(df["quantity"].dropna(), bins=10, edgecolor="black", color="green")

plt.title("Distribution of Order Quantity")
plt.xlabel("Quantity")
plt.ylabel("Frequency")

plt.show()
```

```python
#3 create a histogram of net_amount
```

```python
plt.hist(df["net_amount"].dropna(), bins=10, color="Teal", edgecolor="black")

plt.title("Histogram of Net Amount")
plt.xlabel("Net Amount")
plt.ylabel("Frequency")

plt.show()
```

```python
#4 compare customer age histograms using 5,10 and 20 bins
```

```python
age = df["customer_age"].dropna()

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.hist(age, bins=5, color="green", edgecolor="black")
plt.title("5 Bins")

plt.subplot(1, 3, 2)
plt.hist(age, bins=10, color="green", edgecolor="black")
plt.title("10 Bins")

plt.subplot(1, 3, 3)
plt.hist(age, bins=20, color="green", edgecolor="black")
plt.title("20 Bins")

plt.tight_layout()
plt.show()
```

```python
#5. Display mean and median using plt.axvline()
```

```python
age = df["customer_age"].dropna()

mean_age = age.mean()
median_age = age.median()

plt.hist(age, bins=10, color="green", edgecolor="black")

plt.axvline(mean_age, color="red", linestyle="--", label="Mean")
plt.axvline(median_age, color="blue", label="Median")

plt.title("Customer Age with Mean and Median")
plt.xlabel("Customer Age")
plt.ylabel("Frequency")
plt.legend()

plt.show()
```
