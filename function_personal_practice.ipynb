```python
def maximum(a ,b):
    if a > b:
        return a
        return b
```

```python
def factorial(n):
    fact = 1
    for i in range(1, n+1):
        fact *=i
        return fact
```

```python
factorial(5)
```

```python
def count_vowels(text):
    count = 0
    for ch in text.lower():
        count+=1
        return count
```

```python
count_vowels('python')
```

```python
def reverse(text):
    return text[::-1]
```

```python
reverse('python')
```

```python
def prime(num):
    if num <= 1:
        return False
        for i in range(2,num):
            if num % i == 0:
                return False
                return True
print(prime(13))
```
