```python
def filter_int(a):
    result = list(filter(lambda x:type(x) == int, a))
    return result
data = [10, 'hello',20.5,30,True,'python',40]
```

```python
filter_int(data)
```

```python
def filter_int(*args):
    result = list(filter(lambda x: type(x) == int, args))
    return result
print(filter_int(10,'hello',20.5,40,True,40))
```
