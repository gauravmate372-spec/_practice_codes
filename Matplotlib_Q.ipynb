```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\Admin\Documents\industry_retail_orders_dirty_1000.csv")
```

```python
data = df.groupby(df["order_date"].str[:7])["quantity"].sum()

plt.figure(figsize=(10,5))
plt.plot(data.index, data.values, color="blue", label="Quantity")

plt.title("Monthly Order Quantity")
plt.xlabel("Month")
plt.ylabel("Total Quantity")
plt.xticks(rotation=80, ha="right")
plt.legend()
plt.grid()
plt.show()
```
